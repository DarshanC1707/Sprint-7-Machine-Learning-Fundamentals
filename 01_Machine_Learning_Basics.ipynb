## Complete Machine Learning Workflow

```mermaid
flowchart TD
 A[Define business problem] --> B[Collect data]
 B --> C[Explore and validate data]
 C --> D[Split training and holdout data]
 D --> E[Fit preprocessing on training data]
 E --> F[Train model]
 F --> G[Validate and tune on training folds]
 G --> H[Final evaluation on untouched test set]
 H --> I[Deploy and monitor]
```

**Key rule:** Do not use test-set information to choose features, transformations, hyperparameters, or models. Training and validation support development; the untouched test set estimates final generalization.
